# **Problema 1: Clasificador de temperatura**

Crea un programa que imprima los números desde -30 hasta 50.

Para cada número, además de imprimirlo, el programa debe mostrar un texto según estas reglas:

si el número es menor que 0, mostrar "Muy frío"
si el número está entre 0 y 24, mostrar "Templado"
si el número es mayor o igual a 25, mostrar "Caluroso".


```python
for temperatura in range(-30, 51): #agregamos un número por sobre el límite para cerrar, ya que el comando de "range" lo excluye.
    if temperatura < 0:
        print(temperatura, "Muy frío")
    elif temperatura <= 24:
        print(temperatura, "Templado")
    else:
        print(temperatura, "Caluroso")
```

**Resolución:** Para este ejercicio generé un código simple el cual abarcara los número pedidos, aumentandolo a 51; por la anotación misma del ejercicio.
Asignando el valor de muy frío, templado y caluroso como se pedia, utilizando el texto: "temperatura".




# **Problema 2: Números especiales del robot**
Crea un programa que imprima los números del 1 al 100 que sean divisibles por 5.

Además, el programa debe cumplir estas condiciones:

si el número es divisible por 5, imprimir el número junto con el texto "Bip" si el número también es divisible por 10, imprimir el número junto con el texto "Bip Bip".


```python
for número in range(1, 101):
    if número % 10 == 0:
        print(número, "Bip Bip")
    elif número % 5 == 0:
        print(número, "Bip")
```

**Resolución:** Utilizando la estructura simil al ejercicio anterior (range, print, for/if/elif) es que generamos la cantidad de numeros pedidos. Y para lograr la respuesta de los divisibles se asignó "%" para verficar si esra divisible por 10 para el Bip doble, manteniendo el bip unico en los divisibles de 5.

# **Problema 3: Detector de puntaje gamer**
Crea un programa que solicite al usuario dos números, los sume y luego muestre un mensaje según el resultado.

Reglas

si la suma es menor que 50, mostrar "Nivel principiante"
si la suma es mayor o igual a 50 y menor que 80, mostrar "Nivel intermedio"
si la suma es mayor o igual a 80, mostrar "Nivel pro".

```python
#ingreso de los niveles por el usuario
Puntaje_1 = int(input("Ingrese primer puntaje: "))
Puntaje_2 = int(input("Ingrese segundo puntaje: "))
suma = Puntaje_1 + Puntaje_2

#calculo de las cifras otorgadas
if suma < 50:
    print("Resultado: Nivel principiante")
elif suma < 80:
    print("Resultado: Nivel intermedio")
else:
    print("Resultado: Nivel pro")

#resultado de la suma por su categorización asignada
```

**Resolución:** Haciendo uso de los comandos int e input para este caso, es que asignamos la petición de "Puntaje_1" Y "Puntaje_2" al usuario, al otorgarse esos valores se reliza la suma asignada de ambos, logrando la categorización de los 3 niveles a disposición (como se visualiza al usar el código).


# **Problema 4: Calculador de perfil musical**
Crea un programa que evalúe el perfil del usuario según su edad y sus gustos.

El usuario debe recibir los siguientes parámetros:

edad
le_gusta_musica
le_gusta_bailar
Y debe retornar uno de los siguientes mensajes:

"Eres una estrella de la fiesta" si la persona tiene 18 o más años, le gusta la música y le gusta bailar.

"Te gusta la música, pero no tanto bailar" si le gusta la música pero no le gusta bailar.

"Tienes tu propio estilo" en cualquier otro caso.

```python
edad = int(input("edad: "))
le_gusta_musica = input("¿Te gusta la música? (si/no): ")
le_gusta_bailar = input("¿Te gusta bailar? (si/no): ")

if edad >= 18 and le_gusta_musica == "si" and le_gusta_bailar == "si":
    print("Eres una estrella de la fiesta")
elif le_gusta_musica == "si" and le_gusta_bailar == "no":
    print("Te gusta la música, pero no tanto bailar")
else:
    print("¡Tienes tu propio estilo!")
```

**Resolución**: Para este ejercicio seguí la logica de ejercicios anteriores con el uso de int e input, más las condicionales. Lo que permitió finalmente (en base a multiples respuestas) dar la respuesta de "¡Tienes tu propio estilo!" al ocurrir cualquier otra respuesta a las no vistas; el "Eres una estrella de la fiesta" y "Te gusta la música, pero no tanto bailar" en base a sus respuestas del cuestionario de sí y no.
